```python
import molviewspec as mvs
builder = mvs.create_builder()
structure = builder.download(url='https://files.wwpdb.org/download/1h9t.cif').parse(format='mmcif').assembly_structure()
```

```python
structure.component(selector="polymer").representation(type="cartoon").color(color="red")
```

```python

protein = structure.component_from_uri(uri='./annotations-1h9t.cif', format='cif',
    block_header='1h9t_annotations', category_name='components',
    field_name='component', field_values='Protein', schema='chain')

dna = structure.component_from_uri(uri='./annotations-1h9t.cif', format='cif',
    category_name='components', field_values='DNA', schema='chain')

ions = structure.component_from_uri(uri='./annotations-1h9t.cif', format='cif',
    category_name='components', field_values=['Gold', 'Chloride'],
    schema='chain')
```

```python

structure.label_from_uri(
    uri='./annotations-1h9t.cif', format='cif',
    block_header='1h9t_annotations', category_name='annotations',
    field_name='label', schema='residue_range')
```

```python

builder.save_state(destination='just_structure.mvsj',
           title='An example with MVS annotations', indent=2)
```

```python
import molviewspec as mvs
import zipfile

builder = mvs.create_builder()
structure = builder.component_from_uri("./local_file.cif").parse(format='mmcif').assembly_structure()
structure.representation(type="cartoon").color(color="red")
builder.save_state(destination='from_local.mvsj',title='local_file', indent=2)

with zipfile.ZipFile(".mvsx", mode="w") as z:
    z.write("from_local.mvsj", arcname="index.mvsj")
    z.write("local_file.cif", arcname="local_file.cif")
```
